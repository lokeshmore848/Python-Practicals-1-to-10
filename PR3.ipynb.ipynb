```python
print("---- 1. Conditional Statements ----")
number = 15

if number < 0:
    print(f"{number} is Negative")
elif number == 0:
    print(f"{number} is Zero")
elif number % 2 == 0:
    print(f"{number} is Positive Even")
else:
    print(f"{number} is Positive Odd")

print("\n---- 2. Looping Statements ----")
print("For loop output:")
for item in ["Mango", "Banana", "Cherry"]:
    print(" ", item)

print("\nWhile loop output:")
count = 1
while count <= 3:
    print(f"  Count is: {count}")
    count += 1

print("\n---- 3. Jump / Transfer Statements ----")

print("Loop with 'continue' (skips 3) and 'break' (stops at 6):")
for val in range(1, 10):
    if val == 3:
        continue 
    if val == 6:
        break  
    print(" ", val)

print("\nUsing 'pass' as a placeholder:")
for val in [10, 20]:
    if val == 10:
        pass 
    print(f"  Processed {val}")
```

```python
score = float(input("Enter the student's score (0-100): "))

if score < 0 or score > 100:
    print("Invalid score! Please enter a value between 0 and 100.")
elif score >= 75:
    grade = "A"
    print(f"Grade: {grade} (First Class with Distinction)")
elif score >= 60:
    grade = "B"
    print(f"Grade: {grade} (First Class)")
elif score >= 50:
    grade = "C"
    print(f"Grade: {grade} (Second Class)")
elif score >= 40:
    grade = "D"
    print(f"Grade: {grade} (Pass Class)")
else:
    grade = "E"
    print(f"Grade: {grade} (Fail)")
```

```python
print("---- 1. Using 'break' ----")
for i in range(1, 8):
    if i == 7:
        print("  Breaking at i =", i)
        break
    print("  Value:", i)

print("\n---- 2. Using 'continue' ----")
for i in range(1, 6):
    if i == 4:
        print("  Skipping i =", i)
        continue
    print("  Value:", i)

print("\n---- 3. Using 'pass' ----")
for i in range(1, 6):
    if i == 5:
        pass
        print("  Encountered pass at i =", i)
    print("  Value:", i)
```
