```python
# 1. Write data into a file
file = open("student.txt", "w")
file.write("Student Name: Lokesh More\n")
file.write("Roll Number: 64\n")
file.write("Course: AIML\n")
file.close()

print("Data written successfully.")

# 2. Read data from the file
file = open("student.txt", "r")
print("\nContents of student.txt:")
print(file.read())
file.close()

# 3. Append data to the file
file = open("student.txt", "a")
file.write("Semester: Fifth\n")
file.write("Subject: Python Programming\n")
file.close()

print("Data appended successfully.")

# 4. Read updated contents
file = open("student.txt", "r")
print("\nUpdated contents of student.txt:")
print(file.read())
file.close()

# 5. Copy contents into another file
source = open("student.txt", "r")
destination = open("student_copy.txt", "w")

destination.write(source.read())

source.close()
destination.close()

print("File copied successfully.")

# 6. Display copied file
file = open("student_copy.txt", "r")
print("\nContents of student_copy.txt:")
print(file.read())
file.close()

print("All file operations completed successfully.")
```
