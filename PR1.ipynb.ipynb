```python
# Command Line Argument
import sys

# Simulate command line arguments in Jupyter
sys.argv = ["script.py"]

print("Script Name:", sys.argv[0])
print("Total Arguments Passed:", len(sys.argv) - 1)

if len(sys.argv) > 1:
    print("Arguments List:")
    for index, arg in enumerate(sys.argv[1:], start=1):
        print(f"Argument {index}: {arg}")

else:
    print("\nNo additional arguments were passed.")
```

```python
import sys

def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False

    return True


# Simulate command line arguments in Jupyter
sys.argv = ["script.py", "10", "15", "7", "8", "11", "13"]

if len(sys.argv) > 1:
    # Convert command line arguments to integers
    numbers = [int(arg) for arg in sys.argv[1:]]

    even_numbers = [num for num in numbers if num % 2 == 0]
    odd_numbers = [num for num in numbers if num % 2 != 0]
    prime_numbers = [num for num in numbers if is_prime(num)]

    print("Given Numbers :", numbers)
    print("Even Numbers :", even_numbers)
    print("Odd Numbers :", odd_numbers)
    print("Prime Numbers :", prime_numbers)

else:
    print("Please pass integers as command line arguments.")
```
