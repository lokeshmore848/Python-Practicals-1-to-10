```python
def linear_search(data_list, key):
    for i in range(len(data_list)):
        if data_list[i] == key:
            return i

    return -1


elements = [42, 12, 88, 5, 23, 77, 30]
target = 23

print("Original List:", elements)
print(f"Searching for: {target}")

index = linear_search(elements, target)

if index != -1:
    print(f"Result: Element {target} found at index {index}.")
else:
    print(f"Result: Element {target} not found in the list.")
```

```python
import time

def linear_search(arr, target):
    """Performs linear search and counts comparisons to show impact of list size."""
    comparisons = 0
    for index, val in enumerate(arr):
        comparisons += 1
        if val == target:
            return index, comparisons

    return -1, comparisons


sizes = [1_000, 100_000, 10_000_000]

print(f"{'List Size (n)':<15}{'Target Position':<20}{'Comparisons':<15}{'Time Taken (s)':<15}")
print("-" * 65)

for size in sizes:
    data = list(range(size))
    target = size - 1

    start_time = time.perf_counter()
    index, comparisons = linear_search(data, target)
    elapsed_time = time.perf_counter() - start_time

    print(
        f"{size:<15}{'Last Element':<20}{comparisons:<15}{elapsed_time:.6f}"
    )
```

```python
def search_first_occurrence(arr, target):
    for index, value in enumerate(arr):
        if value == target:
            return index  # Stops at the first match

    return -1


def search_all_occurrences(arr, target):
    indices = []
    for index, value in enumerate(arr):
        if value == target:
            indices.append(index)

    return indices


numbers = [10, 25, 40, 25, 60, 25, 90]
target_val = 25

print("List:", numbers)
print(f"Searching for target: {target_val}\n")

first_idx = search_first_occurrence(numbers, target_val)

if first_idx != -1:
    print(f"Standard Search -> First found at index: {first_idx}")
else:
    print("Standard Search -> Element not found.")

all_idx = search_all_occurrences(numbers, target_val)

if all_idx:
    print(f"All Occurrences Search -> Found at indices: {all_idx}")
    print(f"Total occurrences count: {len(all_idx)}")
else:
    print("All Occurrences Search -> Element not found.")
```

```python
import time

def linear_search_end_element(arr, target):
    comparisons = 0
    found_index = -1

    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == target:
            found_index = i
            break

    return found_index, comparisons


sizes = [10, 10_000, 1_000_000]

print(f"{'List Size (n)':<15}{'Comparisons':<15}{'Time Taken (s)':<15}")
print("-" * 45)

for size in sizes:
    data = list(range(size))
    target = data[-1]  # Target is strictly at the end of the list

    start_time = time.perf_counter()
    idx, comps = linear_search_end_element(data, target)
    elapsed_time = time.perf_counter() - start_time

    print(f"{size:<15}{comps:<15}{elapsed_time:.6f}")
```

```python
import time

def linear_search(arr, target):
    comparisons = 0
    for val in arr:
        comparisons += 1
        if val == target:
            return comparisons
    return comparisons


def binary_search(arr, target):
    comparisons = 0
    low = 0
    high = len(arr) - 1

    while low <= high:
        comparisons += 1
        mid = (low + high) // 2

        if arr[mid] == target:
            return comparisons
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return comparisons


n = 5_000_000
data = list(range(n))
target = n - 1

print(f"Dataset size: {n:,} elements")
print(f"Target element: {target:,} (Worst Case)\n")

start = time.perf_counter()
lin_comps = linear_search(data, target)
lin_time = time.perf_counter() - start

start = time.perf_counter()
bin_comps = binary_search(data, target)
bin_time = time.perf_counter() - start

print(f"{'Algorithm':<18}{'Comparisons':<18}{'Execution Time (s)'}")
print("-" * 52)
print(f"{'Linear Search':<18}{lin_comps:<18}{lin_time:.6f}")
print(f"{'Binary Search':<18}{bin_comps:<18}{bin_time:.6f}")
```
