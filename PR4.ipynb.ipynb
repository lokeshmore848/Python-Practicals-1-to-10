```python
import math

num = float(input("Enter a number: "))
if num >= 0:
    print(f"Square root of {num} is: {math.sqrt(num)}")
else:
    print("Cannot calculate square root of a negative number using math.sqrt().")
```

```python
import math

number = 7.8
result = math.floor(number)

print("Original number:", number)
print("Rounded down integer:", result)
```

```python
import math
import random

rand_num = random.random()

print(f"Generated number: {rand_num}")

try:
    log_value = math.log10(rand_num)
    print(f"Logarithm (base 10): {log_value}")
except ValueError as e:
    print(f"Error: {e}")
```
