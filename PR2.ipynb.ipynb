```python
numbers = [1, 2, 3, 4, 5]
second_element = numbers[1]
third_element = numbers[2]
extracted = [second_element, third_element]

squared_numbers = [second_element**2, third_element**2]

result = squared_numbers + numbers

print("Original List:", numbers)
print("Extracted Elements (2nd and 3rd):", extracted)
print("Squared Numbers List:", squared_numbers)
print("Concatenated Result List:", result)
```

```python
colors = ('red', 'green', 'blue')
first_element = colors[0]

repeated_tuple = (first_element,) * 3

result = repeated_tuple + colors

print("Original Tuple:", colors)
print("Extracted First Element:", first_element)
print("Repeated Tuple:", repeated_tuple)
print("Final Concatenated Tuple:", result)
```

```python
import re

text = "The quick brown fox jumps over the lazy dog"

pattern = r"\b\w+\b"

resulting_string = re.sub(
    pattern,
    lambda m: m.group().upper(), text, flags=re.IGNORECASE
)

print("Original text  :", text)
print("Resulting String:", resulting_string)
```
